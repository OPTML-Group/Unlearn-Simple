```python
# Modify the plot_comparison function to plot multiple x_cols in the same graph (subplot per y_col)

def plot_comparison(df, x_col, y_col):
    import matplotlib.pyplot as plt
    
    # Ensure x_col and y_col are lists for consistency
    if not isinstance(x_col, list):
        x_col = [x_col]
    if not isinstance(y_col, list):
        y_col = [y_col]
    
    # Create subplots: one per y_col (rows), shared x if multiple y
    fig, axes = plt.subplots(len(y_col), 1, figsize=(10, 6 * len(y_col)), sharex=True)
    if len(y_col) == 1:
        axes = [axes]  # Make it a list for consistency
    
    colors = ['skyblue', 'orange', 'green', 'red', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan']
    
    for i, y in enumerate(y_col):
        ax = axes[i]
        for j, x in enumerate(x_col):
            ax.scatter(df[x], df[y], alpha=0.7, color=colors[j % len(colors)], label=x)
        ax.set_xlabel('X Values' if len(x_col) > 1 else x_col[0])
        ax.set_ylabel(y)
        ax.set_title(f'{y} by X Columns')
        ax.tick_params(axis='x', rotation=45)
        if len(x_col) > 1:
            ax.legend()
    
    plt.tight_layout()
    plt.show()
```

```python
def print_df(df):
    df.index = range(1, len(df) + 1)
    df.index.name = '0'
    print(tabulate(df, headers='keys', tablefmt='tsv', showindex=True))
```

```python
# Define a function to compute and display the summary table
def create_summary_table(df, group_by_column, metrics=['retain_vs_all_auc', 'forget_vs_all_auc', 'holdout_vs_all_auc', 'multi_class_auc', 'retain_vs_all_auc_at_1_fp', 'forget_vs_all_auc_at_1_fp', 'holdout_vs_all_auc_at_1_fp']):
    import pandas as pd
    from tabulate import tabulate
        
    # Group by the specified column(s) and compute averages for the selected metrics
    summary_df = df.groupby(group_by_column).agg({m: 'mean' for m in metrics}).reset_index()
    
    # Set index to start from 1 and name it '0'
    summary_df.index = range(1, len(summary_df) + 1)
    summary_df.index.name = '0'
    
    # Display the table in TSV format for easy copy-paste to Google Sheets
    # print_df(summary_df)
    return summary_df
```

```python
def get_last_four_tables(file_path):
    import json
    import pandas as pd
    
    # Load data
    with open(file_path, 'r') as f:
        data = json.load(f)
    
    # Create my_df
    table_data = []
    for entry in data:
        model = entry['model_name']
        method = entry['method']
        binary_comparisons = entry['binary_comparisons']
        
        for rephrasing in entry['rephrasing_results']:
            rephrasing_method = rephrasing['neighbor_method']
            classifier = rephrasing['classifier']
            
            retain_auc = rephrasing['Retain vs All AUC']
            forget_auc = rephrasing['Forget vs All AUC']
            holdout_auc = rephrasing['Holdout vs All AUC']
            multi_auc = rephrasing['Multi-class AUC']
            retain_at_1_fp = rephrasing['Retain_vs_All_auc_at_1_fp']
            forget_at_1_fp = rephrasing['Forget_vs_All_auc_at_1_fp']
            holdout_at_1_fp = rephrasing['Holdout_vs_All_auc_at_1_fp']
            
            accuracy = binary_comparisons['holdout_vs_all'][classifier]['accuracy']
            f1 = binary_comparisons['holdout_vs_all'][classifier]['f1']
            
            table_data.append({
                'model': model,
                'method': method,
                'rephrasing_method_(if exists)': rephrasing_method,
                'classifier': classifier,
                'retain_vs_all_auc': retain_auc,
                'forget_vs_all_auc': forget_auc,
                'holdout_vs_all_auc': holdout_auc,
                'multi_class_auc': multi_auc,
                'accuracy': accuracy,
                'f1': f1,
                'retain_vs_all_auc_at_1_fp': retain_at_1_fp,
                'forget_vs_all_auc_at_1_fp': forget_at_1_fp,
                'holdout_vs_all_auc_at_1_fp': holdout_at_1_fp
            })
            
    my_df = pd.DataFrame(table_data)
    
    # Create theirs_df
    table_data = []
    for entry in data:
        model = entry['model_name']
        method = entry['method']
        
        for baseline_method, baseline_data in entry['baselines'].items():
            if baseline_method == "spv_mia":
                # Handle nested structure for "spv_mia" (max and mean)
                for sub_method, sub_data in baseline_data.items():
                    table_data.append({
                        'model': model,
                        'method': method,
                        'baseline type': f"{baseline_method}_{sub_method}",  # e.g., "spv_mia_max", "spv_mia_mean"
                        'retain_vs_all_auc': sub_data.get('retain_vs_all_auc', None),
                        'forget_vs_all_auc': sub_data.get('forget_vs_all_auc', None),
                        'holdout_vs_all_auc': sub_data.get('holdout_vs_all_auc', None),
                        'multi_class_auc': sub_data.get('multi_class_auc', None),
                        'retain_vs_all_auc_at_1_fp': sub_data.get('retain_vs_all_auc_at_1_fp', None),
                        'forget_vs_all_auc_at_1_fp': sub_data.get('forget_vs_all_auc_at_1_fp', None),
                        'holdout_vs_all_auc_at_1_fp': sub_data.get('holdout_vs_all_auc_at_1_fp', None)
                    })
            else:
                # Handle non-nested baselines normally
                table_data.append({
                    'model': model,
                    'method': method,
                    'baseline type': baseline_method,
                    'retain_vs_all_auc': baseline_data.get('retain_vs_all_auc', None),
                    'forget_vs_all_auc': baseline_data.get('forget_vs_all_auc', None),
                    'holdout_vs_all_auc': baseline_data.get('holdout_vs_all_auc', None),
                    'multi_class_auc': baseline_data.get('multi_class_auc', None),
                    'retain_vs_all_auc_at_1_fp': baseline_data.get('retain_vs_all_auc_at_1_fp', None),
                    'forget_vs_all_auc_at_1_fp': baseline_data.get('forget_vs_all_auc_at_1_fp', None),
                    'holdout_vs_all_auc_at_1_fp': baseline_data.get('holdout_vs_all_auc_at_1_fp', None)
                })
    
    theirs_df = pd.DataFrame(table_data)
    
    # Define create_summary_table function (modified to return without printing)
    def create_summary_table(df, group_by_column, metrics=['retain_vs_all_auc', 'forget_vs_all_auc', 'holdout_vs_all_auc', 'multi_class_auc', 'retain_vs_all_auc_at_1_fp', 'forget_vs_all_auc_at_1_fp', 'holdout_vs_all_auc_at_1_fp']):

        # Group by the specified column(s) and compute averages for the selected metrics
        summary_df = df.groupby(group_by_column).agg({m: 'mean' for m in metrics}).reset_index()
        
        # Multiply the specified metrics by 100 and round to 2 decimal places
        for col in metrics:
            summary_df[col] = (summary_df[col] * 100).round(2)
        
        # Set index to start from 1 and name it '0'
        summary_df.index = range(1, len(summary_df) + 1)
        summary_df.index.name = '0'
        
        # Display the table in TSV format for easy copy-paste to Google Sheets
        # print_df(summary_df)
        return summary_df
    
    # Create the four summary tables
    summary_by_classifier = create_summary_table(my_df, 'classifier')
    summary_by_rephrasing = create_summary_table(my_df, 'rephrasing_method_(if exists)')
    summary_by_baseline = create_summary_table(theirs_df, 'baseline type')
    summary_by_classifier_and_rephrasing = create_summary_table(my_df, ['classifier', 'rephrasing_method_(if exists)'])
    
    ret_dict = {
        'summary_by_classifier': summary_by_classifier,
        'summary_by_rephrasing': summary_by_rephrasing,
        'summary_by_baseline': summary_by_baseline,
        'summary_by_classifier_and_rephrasing': summary_by_classifier_and_rephrasing
    }
    return ret_dict
```

```python
# ...existing code...

# New cell to process all sheets and add params to tables
def process_all_sheets(sheets_names, sheets_tables):
    concatenated_tables = {
        'summary_by_classifier': [],
        'summary_by_rephrasing': [],
        'summary_by_baseline': [],
        'summary_by_classifier_and_rephrasing': []
    }
    base_path = '/home/liranc6/W25/adversarial-attacks-on-deep-learning/project/Unlearn-Simple/results/subset_1000/{}/all_experiment_results.json'
    
    for sheet_name, param_dict in sheets_tables.items():
        file_path = base_path.format(sheet_name)
        try:
            tables = get_last_four_tables(file_path)
        except FileNotFoundError:
            print(f"File not found: {file_path}")
            continue
        
        for table_name, table in tables.items():
            # Add param columns at the beginning
            for param, value in param_dict.items():
                table.insert(0, param, value)
            concatenated_tables[table_name].append(table)

    # Concatenate the lists into DataFrames
    for table_name in concatenated_tables:
        concatenated_tables[table_name] = pd.concat(concatenated_tables[table_name], ignore_index=True)
    
    return concatenated_tables

# Call the function
all_processed_tables = process_all_sheets(sheets_names, sheets_tables)
```
