```python
import os
import json

# load raw_forget
base_data_dir = '/home/liranc6/W25/adversarial-attacks-on-deep-learning/project/Unlearn-Simple/MUSE/data'
file_paths = {
    'forget': os.path.join(base_data_dir, 'news/raw/forget.json'),
    'retain1': os.path.join(base_data_dir, 'news/raw/retain1.json'),
    'holdout': os.path.join(base_data_dir, 'news/raw/holdout.json')
}
privleak_files = {'forget': os.path.join(base_data_dir, 'news/privleak/forget.json'),
                 'retain': os.path.join(base_data_dir, 'news/privleak/retain.json'),
                 'holdout': os.path.join(base_data_dir, 'news/privleak/holdout.json')}
    
raw_data = {}

for file_type, file_path in file_paths.items():
    with open(file_path, 'r', encoding='utf-8') as f:
        raw_data[file_type] = json.load(f)
        
# load privleak
privleak_data = {}
for file_type, file_path in privleak_files.items():
    with open(file_path, 'r', encoding='utf-8') as f:
        privleak_data[file_type] = json.load(f)
    
```

```python
for filetype in privleak_data.keys():
    filetype2 = filetype if filetype != 'retain' else 'retain1'
    for i, item in enumerate(privleak_data[filetype]):
        if item in raw_data[filetype2]:
            print(f"Item {i} found both in privleak and raw data for {filetype}.")
```

```python
retain1_items = list(dataset['retain1'])
retain2_items = list(dataset['retain2'])

in_both = []
for item in privleak_data['retain']:
    if item in retain1_items or item in retain2_items:
        in_both.append(item)
        
    print(f"{len(item)=},\n {item=}")
    break

print(f"Number of privleak retain items found in retain1 or retain2: {len(in_both)}")
if in_both:
    print("Example item:", in_both[0])
```

```python
#find min len of item in privleak_data
splits = ['forget', 'retain', 'holdout']
for split in splits:
    min_len = min(len(item) for item in privleak_data[split])
    max_len = max(len(item) for item in privleak_data[split])
    print(f"Maximum length of items in {split}: {max_len}")
    print(f"Minimum length of items in {split}: {min_len}")
    
```
